## Đề tài: Ứng dụng Học Sâu trong Nhận diện và Phân loại Tổn thương Đường Tiêu hóa từ Ảnh Nội soi
**Học viên thực hiện:** Lê Huy Phát | **Đơn vị:** Đại học Công Thương TP.HCM (HUIT)  
**Mô hình Đề xuất:** Hybrid CNN-CBAM-Transformer (ResNet-50 + CBAM Stage 4 + 2-layer Transformer + GAP Head)  
**Bộ Dữ liệu:** Schema 17 phân lớp y sinh sạch (Zero Video Leakage, Smart Aspect-Ratio Letterbox, Class-Balanced Focal Loss)  
**Mục tiêu Task D48:** Áp dụng 5 trụ cột tối ưu hóa SOTA (Warm Restarts, Differential LR, Regularization Dropout 0.35, Curriculum Augmentation, SWA) để duy trì khả năng hội tụ liên tục suốt 50 epochs, bứt phá mốc **Val Macro F1 $\ge 85\%$**, xuất trọn bộ **24 tệp kết quả lâm sàng** chuẩn y khoa.

---
### 📋 5 Trụ Cột Tối Ưu Hóa Đột Phá Trong Run 6:
1. **Cosine Annealing with Warm Restarts:** $T_0=15, T_{mult}=1$, 3 chu kỳ restart bứt phá cực tiểu cục bộ (local minima).
2. **Layer-wise Differential LR:** Backbone ($10^{-5}$), CBAM + Transformer ($8\times 10^{-5}$), Classifier Head ($2.5\times 10^{-4}$).
3. **Regularization Toàn diện:** Weight Decay $3\times 10^{-2}$, Dropout $0.35$, Label Smoothing $0.10$.
4. **Dynamic Curriculum Augmentation:** Epoch 1-15 nạp nhẹ; Epoch 16-50 kích hoạt CoarseDropout (Cutout 5-10%) & GridDistortion.
5. **Stochastic Weight Averaging (SWA):** Tích lũy trung bình trọng số từ Epoch 25 tìm nghiệm vùng đáy phẳng (Flat Minima).
